### using spark.sql

```python
data=[(1,'alice',5000),(2,'bob',60000),(3,'charlie',45000)]
colums=['id','name','salary']
df=spark.createDataFrame(data, colums)
df.show()
```

```python
df.createOrReplaceTempView("employees")
```

```python
df1=spark.sql('''select * from employees where salary=60000''')
```

```python
df.show() 
df1.show()
```

### using spark.createdataframe using pythin list


```python
from pyspark.sql.types import *

data=[(1,'india'),(2,'japan'),(3,'australia')]

schema=StructType([
    StructField('id',IntegerType(),True),
    StructField('country',StringType(),True)
])

df=spark.createDataFrame(data,schema)
df.show()
```
