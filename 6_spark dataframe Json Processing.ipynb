### explode for array list json

### schema for json

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DateType, BooleanType


customer_schema = StructType([
    StructField("cust_id", IntegerType(), True),
    StructField("cust_nm", StringType(), True),
    StructField("address", StringType(), True),
    StructField("start_dt", DateType(), True),
    StructField("end_dt", DateType(), True),
    StructField("isActive", BooleanType(), True)
])
```

```python
df=spark.read.format("json").schema(customer_schema).option("multiline","true").load("abfss://devcontainer@discoveryprodfixp.dfs.core.windows.net/json files/customers.json")

df.display()
```

```python

from pyspark.sql.types import (
    StructType, StructField,
    StringType, IntegerType,
    ArrayType
)

customer_schema = StructType([
    StructField("cust_id", IntegerType(), True),
    StructField("cust_nm", StringType(), True),

    StructField("address", StructType([
        StructField("street", StringType(), True),
        StructField("city", StringType(), True),
        StructField("pincode", IntegerType(), True)
    ]), True),

    StructField("orders", ArrayType(
        StructType([
            StructField("order_id", StringType(), True),
            StructField("amount", IntegerType(), True),
            StructField("status", StringType(), True)
        ])
    ), True)
])




```

```python
df=spark.read.schema(customer_schema).json("abfss://devcontainer@discoveryprodfixp.dfs.core.windows.net/json files/customers_1.json")
from pyspark.sql.functions import explode

df=df.select("cust_id","cust_nm","address.*",explode("orders").alias("orders"))
df=df.select("cust_id","cust_nm","street","city","pincode","orders.*")
df.display()
```
