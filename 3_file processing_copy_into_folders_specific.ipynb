```python
inbound_path = ("abfss://devcontainer@discoveryprodfixp.dfs.core.windows.net/inbound/")
outbound_path = ("abfss://devcontainer@discoveryprodfixp.dfs.core.windows.net/outbound/")
```

```python
# List all files in the inbound path
files = dbutils.fs.ls(inbound_path)

# Move each file to the appropriate subfolder in outbound based on its name prefix
for file in files:
    if not file.isDir():
        file_name = file.name
        # Determine the folder name based on the file name prefix (e.g., "order_123.csv" -> "orders")
        prefix = file_name.split("_")[0].lower()  # extract prefix before the first underscore
        folder_name = prefix + "s" if not prefix.endswith("s") else prefix  # pluralize for folder name

        destination = f"{outbound_path}{folder_name}/{file_name}"
        print(f"Moving: {file_name} -> {destination}")
        dbutils.fs.cp(file.path, destination)

print("\nAll files have been moved to their respective folders in outbound.")
```
