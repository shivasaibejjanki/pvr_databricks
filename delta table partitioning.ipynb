### static partitioning

```python
data = [
    (1, "Ramesh", "IND", "Hyd", 5000),
    (2, "Sita", "IND", "Hyd", 6000),
    (3, "Mohan", "IND", "Blr", 7000),
    (4, "John", "USA", "NY", 8000),
    (5, "David", "USA", "SF", 9000),
    (6, "Mary", "USA", "NY", 7500),
    (7, "Kiran", "IND", "Chn", 6500),
    (8, "Anu", "IND", "Blr", 7200),
    (9, "Sam", "UK", "LON", 8200),
    (10, "Peter", "UK", "LON", 8300)
]

columns = ["emp_id", "emp_name", "country", "city", "salary"]

df = spark.createDataFrame(data, columns)
```

```python
df.write.format("Delta").mode("overwrite").option("overwriteSchema", "true").partitionBy("country","city").saveAsTable("shivasai_catalog.bronze.static_partitioning")
```

```python
%sql
INSERT INTO shivasai_catalog.bronze.static_partitioning PARTITION (country = 'UK') VALUES (11, 'shivasai', 'LON', 150000)
```
